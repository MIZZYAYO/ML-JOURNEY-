```python
# Normal Python code

numbers = [1, 2, 3, 4]
doubled = []
for n in numbers:
    doubled.append(n * 2)    # doubling numbers in the list 

print(doubled)
```

```python
# Using Numpy 

import numpy as np
numbers = np.array([1, 2, 3, 4])
doubled = numbers * 2

print(doubled)
print(type(doubled))  # shows it used numpy 
```

No loop — NumPy applies the operation to every element automatically. This is called vectorization, and it's the core idea behind almost everything NumPy does.

```python
import numpy as np

# 1D array - like a single list
a = np.array([1, 2, 3])

# 2D array - like a grid/table (rows and columns)
b = np.array([[1, 2, 3], [4, 5, 6]])

print(a)
print(a.shape)  

print(b)
print(b.shape) # 2 by 3 - 2 rows and 3 columns 
```

```python
b = np.array([[1, 2, 3], [4, 5, 6]])

print(b[0])        # first row
print(b[1])        # second row
print(b[0, 2])      # row 0, column 2
print(b[:, 1])      # every row, column 1 only   
```

```python

# Useful Array operations  

arr = np.array([10, 20, 30, 40, 50])

print(arr.sum())     # total
print(arr.mean())    # average
print(arr.max())     # largest value
print(arr.min())     # smallest value
print(arr.std())     # standard deviation (spread of the data)

print(arr > 25)       # boolean array: True/False for each element
print(arr[arr > 25])  # filters: only values greater than 25
```

```python
# Pandas and what it looks like 

import pandas as pd

data = {
    "name": ["Ada", "Daniel", "Bob"],
    "balance": [100, 200, 50],
    "active": [True, True, False]
}

df = pd.DataFrame(data)
print(df)
```

```python
data1 = {
    "name": ["Ada", "Daniel", "Bob"],
    "balance": [100, 200, 50],
    "active": [True, True, False]
}
df = pd.DataFrame(data1)

# Selecting a single column
print(df["balance"])

# Selecting multiple columns
print(df[["name", "balance"]])

# Filtering rows (same boolean idea as NumPy)
print(df[df["balance"] > 75])

# Sorting
print(df.sort_values("balance", ascending=False))  # descending order 
```

```python
# Reading real data from a CSV

df = pd.read_csv("example.csv")
print(df.head())    # first 5 rows
print(df.info())    # column types, missing values, row count
print(df.describe())  # quick stats: mean, min, max, etc for numeric columns
```

```python
# Handling missing data

# Check for missing values
print(df.isna())          # True/False grid — True where a value is missing
print(df.isna().sum())    # count of missing values, per column — much more useful in practice

#Dropping missing data
df_dropped = df.dropna()          # removes any row that has at least one missing value

# Filling missing data
df_filled = df.fillna(0)          # replaces every missing value with 0
```

```python
# Grouping and Aggregating 

data = {
    "department": ["Sales", "Sales", "Engineering", "Engineering"],
    "salary": [50000, 60000, 90000, 95000]
}

df2 = pd.DataFrame(data)


# Using Groupby 

print(df2.groupby("department")["salary"].mean())   # average salary per department
print(df2.groupby("department")["salary"].sum())    # total salary per department
```
